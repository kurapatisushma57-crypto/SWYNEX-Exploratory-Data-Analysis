```python

import pandas as pd

# Load the uploaded dataset
df = pd.read_csv("cleaned_dataset.csv")

# Display dataset size
print("Number of rows and columns:", df.shape)

# Display first 5 rows
print("\nFirst 5 rows:")
display(df.head())

# Check column names and data types
print("\nDataset information:")
df.info()

# Check missing values
print("\nMissing values in each column:")
print(df.isnull().sum())

# Check duplicate rows
print("\nNumber of duplicate rows:", df.duplicated().sum())

# Calculate statistical summary
print("\nStatistical summary:")
display(df.describe())
```

```python

numeric_columns = [
    "age",
    "fnlwgt",
    "education_num",
    "capital_gain",
    "capital_loss",
    "hours_per_week"
]

print("MEAN (AVERAGE)")
display(df[numeric_columns].mean().round(2))

print("MEDIAN")
display(df[numeric_columns].median())

print("MINIMUM VALUES")
display(df[numeric_columns].min())

print("MAXIMUM VALUES")
display(df[numeric_columns].max())

print("INCOME DISTRIBUTION")
display(df["income"].value_counts())

print("EDUCATION DISTRIBUTION")
display(df["education"].value_counts().head(10))

print("WORKCLASS DISTRIBUTION")
display(df["workclass"].value_counts().head(10))
```

```python

import matplotlib.pyplot as plt

# Chart 1: Income distribution
df["income"].value_counts().plot(
    kind="bar",
    title="Income Distribution"
)
plt.xlabel("Income Group")
plt.ylabel("Number of People")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()

# Chart 2: Education distribution
df["education"].value_counts().plot(
    kind="bar",
    title="Education Level Distribution",
    figsize=(10, 5)
)
plt.xlabel("Education Level")
plt.ylabel("Number of People")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()

# Chart 3: Working hours distribution
df["hours_per_week"].plot(
    kind="hist",
    bins=20,
    title="Distribution of Weekly Working Hours",
    figsize=(8, 5)
)
plt.xlabel("Hours Worked Per Week")
plt.ylabel("Number of People")
plt.tight_layout()
plt.show()
```

```python

# Count income groups for each education level
education_income = pd.crosstab(
    df["education"],
    df["income"]
)

print("Education Level vs Income")
display(education_income)

# Calculate the percentage earning >50K
high_income_percentage = (
    df.assign(high_income=df["income"].eq(">50K"))
      .groupby("education")["high_income"]
      .mean()
      .mul(100)
      .round(2)
      .sort_values(ascending=False)
)

print("Percentage earning >50K by education level")
display(high_income_percentage)
```

```python

# Calculate percentage earning >50K for each education level
high_income_percentage = (
    df.assign(high_income=df["income"].eq(">50K"))
      .groupby("education")["high_income"]
      .mean()
      .mul(100)
      .sort_values(ascending=False)
)

# Plot the top 10 education levels
high_income_percentage.head(10).sort_values().plot(
    kind="barh",
    figsize=(9, 6),
    title="Top 10 Education Levels by Percentage Earning >50K"
)

plt.xlabel("People Earning >50K (%)")
plt.ylabel("Education Level")
plt.tight_layout()
plt.show()
```

```python

# Check unusually low or high working hours
print("People working more than 60 hours per week:")
print((df["hours_per_week"] > 60).sum())

print("\nPeople working 1 hour or less per week:")
print((df["hours_per_week"] <= 1).sum())

print("\nPeople aged below 18:")
print((df["age"] < 18).sum())

print("\nPeople aged above 70:")
print((df["age"] > 70).sum())

print("\nPeople with capital gain above 10,000:")
print((df["capital_gain"] > 10000).sum())

print("\nPeople with capital loss above 2,000:")
print((df["capital_loss"] > 2000).sum())

print("\nHighest 10 weekly working-hour values:")
display(df["hours_per_week"].value_counts().sort_index().tail(10))
```

```python

# 1. Percentage in each income group
print("INCOME PERCENTAGES")
display(
    (df["income"].value_counts(normalize=True) * 100).round(2)
)

# 2. Percentage in each workclass
print("\nWORKCLASS PERCENTAGES")
display(
    (df["workclass"].value_counts(normalize=True) * 100)
    .round(2)
    .head(10)
)

# 3. Percentage working more than 40 hours per week
print("\nWORKING MORE THAN 40 HOURS")
display(
    round((df["hours_per_week"] > 40).mean() * 100, 2)
)

# 4. Percentage working exactly 40 hours per week
print("\nWORKING EXACTLY 40 HOURS")
display(
    round((df["hours_per_week"] == 40).mean() * 100, 2)
)

# 5. Average weekly hours by income group
print("\nAVERAGE HOURS BY INCOME")
display(
    df.groupby("income")["hours_per_week"].mean().round(2)
)
```

```python

print("Records working more than 60 hours:",
      (df["hours_per_week"] > 60).sum())

print("Records working 1 hour or less:",
      (df["hours_per_week"] <= 1).sum())

print("Records aged above 70:",
      (df["age"] > 70).sum())

print("Records with capital gain above 10,000:",
      (df["capital_gain"] > 10000).sum())

print("Records with capital loss above 2,000:",
      (df["capital_loss"] > 2000).sum())

print("Records working 99 hours per week:",
      (df["hours_per_week"] == 99).sum())
```

```python

import os

os.makedirs("results", exist_ok=True)

# Save income distribution
income_summary = df["income"].value_counts().rename_axis(
    "income"
).reset_index(name="count")
income_summary.to_csv("results/income_distribution.csv", index=False)

# Save education and income analysis
education_income.to_csv("results/education_vs_income.csv")

# Save percentage earning >50K by education
high_income_percentage.rename(
    "percent_earning_over_50K"
).to_csv("results/high_income_by_education.csv")

# Save numerical summary
df.describe().to_csv("results/numerical_summary.csv")

# Save anomaly counts
anomaly_summary = pd.DataFrame({
    "check": [
        "Working more than 60 hours",
        "Working 1 hour or less",
        "Age above 70",
        "Capital gain above 10000",
        "Capital loss above 2000",
        "Working exactly 99 hours"
    ],
    "count": [1110, 20, 539, 770, 325, 85]
})
anomaly_summary.to_csv("results/anomaly_summary.csv", index=False)

print("Analysis results saved successfully!")
print(os.listdir("results"))
```
